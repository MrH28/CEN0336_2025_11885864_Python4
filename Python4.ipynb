````python
# 6. Fatorial de 1000 com while
factorial_1000 = 1
number = 2
while number <= 1000:
    factorial_1000 *= number
    number += 1
print("6:", factorial_1000)

numbers = [101, 2, 15, 22, 95, 33, 2, 27, 72, 15, 52]
````

````python
# 7. Valores pares
print("7:", end=" ")
for number in numbers:
    if number % 2 == 0:
        print(number, end=" ")
print()
````

````python

# 8. Lista ordenada, valores e somas acumulativas
sorted_numbers = sorted(numbers)
even_sum = 0
odd_sum = 0
for number in sorted_numbers:
    print(number, end=" ")
    if number % 2 == 0:
        even_sum += number
    else:
        odd_sum += number
print()
print(f"Soma dos números pares: {even_sum}")
print(f"Soma dos números ímpares: {odd_sum}")
````

````python
# Desafio 1. Fisher-Yates shuffle
import random

shuffled = list("ATGCCCGGCCCGGC")
for index in range(len(shuffled) - 1, 0, -1):
    other_index = random.randrange(index + 1)
    shuffled[index], shuffled[other_index] = shuffled[other_index], shuffled[index]
print("Desafio 1:", "".join(shuffled))
````

````python
# Desafio 2. Identidade entre duas sequências alinhadas
aligned_one = "ATGCCCGGCCCGGC"
aligned_two = "ATGCCCGACCCGGC"
if len(aligned_one) != len(aligned_two):
    raise ValueError("As sequências alinhadas devem ter o mesmo comprimento.")
matching_positions = sum(base_one == base_two for base_one, base_two in zip(aligned_one, aligned_two))
identity = 100 * matching_positions / len(aligned_one)
print(f"Desafio 2: {identity:.2f}% de identidade")
````
