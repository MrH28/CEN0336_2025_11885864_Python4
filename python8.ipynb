## 0. Leitura do arquivo multi-FASTA
Lê o nome do arquivo a partir da entrada do usuário (Enter usa `python8_fasta.txt`) e guarda as sequências em um dicionário `{nome: sequencia}`.

```python
from pathlib import Path

def ler_fasta(nome_arquivo):
    """Lê um arquivo multi-FASTA e retorna um dicionário {nome: sequencia}."""
    sequencias = {}
    nome = None
    with open(nome_arquivo, encoding='utf-8') as fh:
        for linha in fh:
            linha = linha.strip()
            if not linha:
                continue
            if linha.startswith('>'):
                nome = linha[1:].split()[0]
                sequencias[nome] = []
            else:
                sequencias[nome].append(linha.upper())
    return {nome: ''.join(partes) for nome, partes in sequencias.items()}


arquivo = 'python8_fasta.txt'
PASTA_RESULTADO = Path('resultado')
PASTA_RESULTADO.mkdir(exist_ok=True)
seqs_fasta = ler_fasta(arquivo)
def codons(seq):
    return [seq[i:i + 3] for i in range(0, len(seq) - 2, 3)]

with open(PASTA_RESULTADO / 'Python_08.codons-frame-1.nt', 'w', encoding='utf-8') as out:
    for nome, seq in seqs_fasta.items():
        out.write(f'>{nome}-frame-1-codons\n')
        out.write(' '.join(codons(seq)) + '\n')

print(f'{len(seqs_fasta)} sequências lidas de {arquivo}')
print(f'Escrito: {PASTA_RESULTADO / "Python_08.codons-frame-1.nt"}')
```

### Teste com um conjunto de dados pequeno
Sequências cuja resposta conhecemos, para conferir as funções antes de rodar no arquivo real.

```python
teste = {'geneA': 'ATGCATGCT', 'geneB': 'TTTTTAGGCC'}
# respostas esperadas:
# geneA: A=2 T=3 G=2 C=2   códons quadro 1: ATG CAT GCT -> M H A
# geneB: A=1 T=5 G=2 C=2
```

## 1. Composição de nucleotídeos
Estrutura: `seqs[nomeGene][nucleotideo] = contagem`

```python
def composicao(sequencias):
    seqs = {}
    for nome, seq in sequencias.items():
        seqs[nome] = {}
        for nt in seq:
            seqs[nome][nt] = seqs[nome].get(nt, 0) + 1
    return seqs


def imprimir_composicao(seqs):
    for nome in seqs:
        contagens = [str(seqs[nome].get(nt, 0)) for nt in 'ATGC']
        print(nome + '\t' + '\t'.join(contagens))


print('--- teste ---')
imprimir_composicao(composicao(teste))

print('--- arquivo ---')
seqs = composicao(seqs_fasta)
imprimir_composicao(seqs)
```

## 2. Códons no primeiro quadro de leitura
Saída em `Python_08.codons-frame-1.nt`.

```python
def codons(seq, quadro=0):
    """Divide a sequência em códons completos a partir do quadro informado."""
    return [seq[i:i + 3] for i in range(quadro, len(seq) - 2, 3)]


print(codons(teste['geneA']))   # esperado: ['ATG', 'CAT', 'GCT']

with open(PASTA_RESULTADO / 'Python_08.codons-frame-1.nt', 'w', encoding='utf-8') as out:
    for nome, seq in seqs_fasta.items():
        out.write(f'>{nome}-frame-1-codons\n')
        out.write(' '.join(codons(seq, 0)) + '\n')
print(f'Arquivo gerado: {PASTA_RESULTADO / "Python_08.codons-frame-1.nt"}')
```

## 3. Códons nos três primeiros quadros de leitura
Saída em `Python_08.codons-3frames.nt`.

```python
with open(PASTA_RESULTADO / 'Python_08.codons-3frames.nt', 'w') as out:
    for nome, seq in seqs_fasta.items():
        for quadro in range(3):
            out.write(f'{nome}-frame-{quadro + 1}-codons\n')
            out.write(' '.join(codons(seq, quadro)) + '\n')
print(f'Escrito: {PASTA_RESULTADO / "Python_08.codons-3frames.nt"}')
```

## 4. Complemento reverso e seis quadros de leitura
Quadros 1-3: fita direta; quadros 4-6: complemento reverso. O arquivo `Python_08.codons-6frames.nt` é gravado na seção 5, junto com as traduções.

```python
complemento = {'A': 'T', 'T': 'A', 'G': 'C', 'C': 'G'}

def complemento_reverso(seq):
    return ''.join(complemento.get(nt, 'N') for nt in reversed(seq))


def seis_quadros(seq):
    """Retorna um dicionário {numero_do_quadro: lista_de_codons}."""
    rc = complemento_reverso(seq)
    quadros = {}
    for q in range(3):
        quadros[q + 1] = codons(seq, q)
    for q in range(3):
        quadros[q + 4] = codons(rc, q)
    return quadros


print(complemento_reverso('ATGCATGCT'))   # esperado: AGCATGCAT
```

## 5. Tradução, peptídeo mais longo (M => Stop) e ORF correspondente
Gera os quatro arquivos em formato FASTA:
- `Python_08.codons-6frames.nt` — os seis quadros de códons
- `Python_08.translated.aa` — tradução dos seis quadros
- `Python_08.translated-longest.aa` — peptídeo mais longo de cada sequência
- `Python_08.orf-longest.nt` — códons que produziram esse peptídeo

```python
tabela_de_traducao = {
    'GCT':'A', 'GCC':'A', 'GCA':'A', 'GCG':'A',
    'CGT':'R', 'CGC':'R', 'CGA':'R', 'CGG':'R', 'AGA':'R', 'AGG':'R',
    'AAT':'N', 'AAC':'N',
    'GAT':'D', 'GAC':'D',
    'TGT':'C', 'TGC':'C',
    'CAA':'Q', 'CAG':'Q',
    'GAA':'E', 'GAG':'E',
    'GGT':'G', 'GGC':'G', 'GGA':'G', 'GGG':'G',
    'CAT':'H', 'CAC':'H',
    'ATT':'I', 'ATC':'I', 'ATA':'I',
    'TTA':'L', 'TTG':'L', 'CTT':'L', 'CTC':'L', 'CTA':'L', 'CTG':'L',
    'AAA':'K', 'AAG':'K',
    'ATG':'M',
    'TTT':'F', 'TTC':'F',
    'CCT':'P', 'CCC':'P', 'CCA':'P', 'CCG':'P',
    'TCT':'S', 'TCC':'S', 'TCA':'S', 'TCG':'S', 'AGT':'S', 'AGC':'S',
    'ACT':'T', 'ACC':'T', 'ACA':'T', 'ACG':'T',
    'TGG':'W',
    'TAT':'Y', 'TAC':'Y',
    'GTT':'V', 'GTC':'V', 'GTA':'V', 'GTG':'V',
    'TAA':'*', 'TGA':'*', 'TAG':'*'
}


def traduzir(lista_codons):
    # códons desconhecidos (ex.: com N) viram 'X'
    return ''.join(tabela_de_traducao.get(c, 'X') for c in lista_codons)


def peptideo_mais_longo(proteina):
    """Retorna (inicio, fim) do peptídeo mais longo que começa em M e termina
    em um stop (*), com o stop incluído. (None, None) se não houver nenhum."""
    melhor = (None, None)
    inicio = None
    for i, aa in enumerate(proteina):
        if aa == 'M' and inicio is None:
            inicio = i                      # primeiro M depois do último stop
        elif aa == '*':
            if inicio is not None:
                if melhor[0] is None or (i + 1 - inicio) > (melhor[1] - melhor[0]):
                    melhor = (inicio, i + 1)
            inicio = None
    return melhor


print(traduzir(codons(teste['geneA'])))           # esperado: MHA
print(peptideo_mais_longo('AMKLM*GGMW*MAAAAAA'))   # esperado: (1, 6) -> 'MKLM*'
```

```python
with open(PASTA_RESULTADO / 'Python_08.codons-6frames.nt', 'w') as f_codons, \
     open(PASTA_RESULTADO / 'Python_08.translated.aa', 'w') as f_trad, \
     open(PASTA_RESULTADO / 'Python_08.translated-longest.aa', 'w') as f_pep, \
     open(PASTA_RESULTADO / 'Python_08.orf-longest.nt', 'w') as f_orf:

    sem_orf = 0
    for nome, seq in seqs_fasta.items():
        melhor = None   # (tamanho, quadro, peptideo, codons_do_orf)

        for quadro, lista_codons in seis_quadros(seq).items():
            proteina = traduzir(lista_codons)

            f_codons.write(f'>{nome}-frame-{quadro}-codons\n')
            f_codons.write(' '.join(lista_codons) + '\n')

            f_trad.write(f'>{nome}-frame-{quadro}-aa\n')
            f_trad.write(proteina + '\n')

            ini, fim = peptideo_mais_longo(proteina)
            if ini is not None and (melhor is None or fim - ini > melhor[0]):
                melhor = (fim - ini, quadro, proteina[ini:fim], lista_codons[ini:fim])

        if melhor is None:
            # nenhum quadro tem M seguido de stop: registra a sequência mesmo assim
            sem_orf += 1
            f_pep.write(f'>{nome}-no-ORF len=0\n\n')
            f_orf.write(f'>{nome}-no-ORF len=0\n\n')
            continue
        tamanho, quadro, peptideo, orf = melhor
        f_pep.write(f'>{nome}-frame-{quadro}-longest len={tamanho}\n{peptideo}\n')
        f_orf.write(f'>{nome}-frame-{quadro}-orf-longest len={tamanho * 3}\n{"".join(orf)}\n')

print('Escritos: Python_08.codons-6frames.nt, Python_08.translated.aa,')
print('          Python_08.translated-longest.aa, Python_08.orf-longest.nt')
print(f'Sequências sem nenhum peptídeo M=>Stop (registradas como no-ORF): {sem_orf}')
```

### Conferência
O ORF (nt) traduzido deve ser exatamente o peptídeo mais longo.

```python
pep = ler_fasta(PASTA_RESULTADO / 'Python_08.translated-longest.aa')
orf = ler_fasta(PASTA_RESULTADO / 'Python_08.orf-longest.nt')

def id_sequencia(nome):
    return nome.split('-frame-', 1)[0].split('-no-ORF', 1)[0]


# Compare records by original sequence ID, not by FASTA header or order.
pep_por_id = {id_sequencia(nome): sequencia for nome, sequencia in pep.items()}
orf_por_id = {id_sequencia(nome): sequencia for nome, sequencia in orf.items()}
ids_pep = set(pep_por_id)
ids_orf = set(orf_por_id)
ids_comuns = ids_pep & ids_orf
erros = {
    nome: (traduzir(codons(orf_por_id[nome])), pep_por_id[nome])
    for nome in ids_comuns
    if traduzir(codons(orf_por_id[nome])) != pep_por_id[nome]
}
ok = ids_pep == ids_orf and not erros
print(f'{len(pep)} peptídeos; ORFs conferem com os peptídeos:', ok)
if ids_pep != ids_orf:
    print('IDs ausentes:', sorted(ids_pep ^ ids_orf))
if erros:
    print(f'Inconsistências encontradas: {len(erros)}')
    for nome, (traducao, esperado) in list(erros.items())[:3]:
        print(nome, traducao[:60], esperado[:60])
for n in list(pep_por_id)[:3]:
    print(n, pep_por_id[n][:60] + ('...' if len(pep_por_id[n]) > 60 else ''))
```
