## Extra: Expandindo um exercício do Conjunto de Problemas 5 sobre composição de nucleotídeos

  - itere sobre cada linha neste arquivo (seqName\tsequence\n)

         - calcule e armazene a contagem de cada caractere de nucleotídeo único em um dicionário
         - relate o nome, o total de cada contagem de nucleotídeos e o conteúdo de GC


```python
from pathlib import Path
from urllib.request import urlopen

ROOT = Path.cwd()
DATA = ROOT / "python6_data"
DATA.mkdir(exist_ok=True)
URLS = {
    "Python_06.txt": "https://raw.githubusercontent.com/labbces/cen0336/master/files/Python_06.txt",
    "Python_06.seq.txt": "https://raw.githubusercontent.com/labbces/cen0336/master/files/Python_06.seq.txt",
    "Python_06.fastq": "https://raw.githubusercontent.com/labbces/cen0336/master/files/Python_06.fastq",
}

def get_file(name):
    path = DATA / name
    if not path.exists():
        with urlopen(URLS[name], timeout=30) as response:
            path.write_bytes(response.read())
    return path

def reverse_complement(sequence):
    return sequence.translate(str.maketrans("ACGTNacgtn", "TGCANtgcan"))[::-1]

def read_tab_sequences(path):
    records = {}
    with path.open(encoding="utf-8") as handle:
        for line_number, line in enumerate(handle, 1):
            fields = line.strip().split("\t")
            if len(fields) != 2:
                raise ValueError(f"{path.name}: invalid line {line_number}")
            records[fields[0]] = fields[1].upper()
    return records

def parse_fasta(path):
    records, name, parts = {}, None, []
    with path.open(encoding="utf-8") as handle:
        for line in handle:
            line = line.strip()
            if not line:
                continue
            if line.startswith(">"):
                if name is not None:
                    records[name] = "".join(parts)
                name, parts = line[1:].split()[0], []
            elif name is None:
                raise ValueError("FASTA sequence found before header")
            else:
                parts.append(line)
    if name is not None:
        records[name] = "".join(parts)
    return records

# 1-2. Uppercase text and save it.
text_file = get_file("Python_06.txt")
uppercase = [line.upper() for line in text_file.read_text(encoding="utf-8").splitlines()]
print("1. Uppercase text:")
print("\n".join(uppercase))
uppercase_file = DATA / "Python_06_uc.txt"
uppercase_file.write_text("\n".join(uppercase) + "\n", encoding="utf-8")
print("2. Saved:", uppercase_file)

# 3. Reverse complements in FASTA format and save redirect-equivalent output.
seq_file = get_file("Python_06.seq.txt")
records = read_tab_sequences(seq_file)
reverse_file = DATA / "Python_06_reverse_complements.fasta"
with reverse_file.open("w", encoding="utf-8") as output:
    for name, sequence in records.items():
        reverse = reverse_complement(sequence)
        print(f">{name} reverse_complement")
        print(reverse)
        output.write(f">{name} reverse_complement\n{reverse}\n")
print("3. Saved:", reverse_file)

# 4. FASTQ statistics.
fastq_lines = get_file("Python_06.fastq").read_text(encoding="utf-8").splitlines()
line_count = len(fastq_lines)
character_count = sum(map(len, fastq_lines))
print(f"4. Lines: {line_count}")
print(f"4. Characters: {character_count}")
print(f"4. Average line length: {character_count / line_count:.2f}")

# 5. FASTA parser, including multiline records.
fasta_file = ROOT / "Python_06.fasta"
fasta_dict = parse_fasta(fasta_file) if fasta_file.exists() else records
print(f"5. FASTA records: {len(fasta_dict)}")
for name, sequence in fasta_dict.items():
    print(f"{name}: {len(sequence)} bases")

# 6. BioMart gene-set comparisons when TSV files are present.
def gene_ids(path):
    result = set()
    with path.open(encoding="utf-8") as handle:
        for row, line in enumerate(handle):
            value = line.rstrip("\n").split("\t")[0].strip()
            if value and not (row == 0 and value.lower() in {"gene", "gene stable id", "gene_id"}):
                result.add(value)
    return result

gene_paths = {"all": ROOT / "alpaca_all_genes.tsv", "stem": ROOT / "alpaca_stemcellproliferation_genes.tsv", "pigmentation": ROOT / "alpaca_pigmentation_genes.tsv", "tf": ROOT / "alpaca_transcriptionFactors.tsv"}
genes = {key: gene_ids(value) for key, value in gene_paths.items() if value.exists()}
if {"all", "stem"} <= genes.keys(): print("6A. Not stem-cell proliferation:", len(genes["all"] - genes["stem"]))
if {"stem", "pigmentation"} <= genes.keys(): print("6B. Stem-cell and pigmentation:", len(genes["stem"] & genes["pigmentation"]))
if {"tf", "stem"} <= genes.keys(): print("6C. TFs for proliferation:", len(genes["tf"] & genes["stem"]))
missing = [path.name for path in gene_paths.values() if not path.exists()]
if missing: print("6. Missing optional BioMart files:", ", ".join(missing))

# Extra. Nucleotide counts and GC content for every sequence.
print("Extra. Nucleotide composition:")
for name, sequence in records.items():
    counts = {base: sequence.count(base) for base in "ACGTN" if base in sequence}
    gc = (counts.get("G", 0) + counts.get("C", 0)) / len(sequence) if sequence else 0
    print(f"{name}: length={len(sequence)}, counts={counts}, GC={gc:.2%}")
```
